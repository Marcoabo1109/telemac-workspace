## Séries de demonstração sintéticas
Servem para testar lag e permanência. Para medições, importe CSV canônico date,value e confirme unidade, estação e qualidade.

```python
import numpy as np, pandas as pd
from scripts.tapajos_analysis import report
rng=np.random.default_rng(12)
t=pd.date_range('2020-01-01', periods=180)
q=1000+100*rng.normal(size=len(t))
out=ROOT/'reports/synthetic';out.mkdir(parents=True,exist_ok=True)
pd.DataFrame({'date':t,'value':q}).to_csv(out/'up.csv',index=False)
pd.DataFrame({'date':t+pd.Timedelta(days=4),'value':q}).to_csv(out/'down.csv',index=False)
result=report(out/'up.csv',out/'down.csv',out)
result['best']
```
